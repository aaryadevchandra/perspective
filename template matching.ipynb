```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("sift1.jpg")
```

```python
img.shape
```

```python
random_pt = [np.random.randint(100, img.shape[0]-200), np.random.randint(100, img.shape[1])-200]
window_size = 50
```

```python
img_slice = img[random_pt[0]:random_pt[0] + window_size, random_pt[1]: random_pt[1] + window_size]
```

```python
plt.imshow(img_slice)
```

```python
def distance_metric(img1, img2):
    return np.sum(((img1 - img2) ** 2).flatten())
```

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[1, 1], [2, 2]])

sq_error = (A - B) ** 2
print(sq_error)
```

```python
np.sum(sq_error.flatten())
```

```python
min_val = np.inf
sol_slice = None
for j in range(img.shape[0] - window_size + 1):
    distance = distance_metric(img_slice, img[j:j+window_size, random_pt[0]:random_pt[0] + window_size])
    if distance < min_val:
        min_val = distance
        sol_slice = img[j:j+window_size, random_pt[0]:random_pt[0] + window_size]
        print(f"min distance is {distance}")
```

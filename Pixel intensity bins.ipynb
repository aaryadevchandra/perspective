```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("bird.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
flattened_img = img.flatten().astype(dtype=np.uint8)
unique_pixel_vals = list(set([int(pixel_val) for pixel_val in flattened_img]))
hist = {}
for pixel_val in unique_pixel_vals:
    hist[f"{pixel_val}"] = int(np.sum(flattened_img == pixel_val))
    
    
```

```python
hist
```

```python
plt.bar(list(hist.keys()), list(hist.values()), width=.5)
plt.title("Pixel Intensity Histogram")
plt.grid()
```
